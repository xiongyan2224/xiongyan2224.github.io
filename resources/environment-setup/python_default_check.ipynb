# Part 1. Install Packages from the Terminal

Install **NumPy**, **pandas**, and **Matplotlib** using your existing Python installation. Prefer the Python environment you already use for this course. **pathlib** comes with Python and does not need installation.

**Before you begin:** install Python. If you use VS Code, install the Microsoft **Python** and **Jupyter** extensions. The commands below also install **ipykernel**, which allows Python to run notebook cells.

## 1. Open a terminal

- **Windows:** open Command Prompt or PowerShell.
- **macOS:** open Terminal.
- **Linux:** open your terminal application.
- **VS Code, any system:** choose **Terminal → New Terminal**.

Run the commands below in the **terminal**, not in a notebook code cell. Choose the instructions for your operating system.

## 2. Install the packages

### Windows — Command Prompt or PowerShell

```bat
py -m pip install numpy pandas matplotlib ipykernel
```

If `py` is not recognized but `python` works, use:

```bat
python -m pip install numpy pandas matplotlib ipykernel
```

### macOS / Linux

```bash
python3 -m pip install numpy pandas matplotlib ipykernel
```

Wait until installation finishes. `Requirement already satisfied` means the package is already installed. Terminal commands do **not** use `%` before pip.

## 3. Check that the terminal and notebook use the same Python

If you have multiple Python installations, a terminal command may target a different interpreter. In VS Code, open this `.ipynb` file and use **Select Kernel** at the top right → **Python Environments** to choose the Python where you installed the packages. In JupyterLab, select the corresponding Python kernel. To check your notebook's interpreter, run this short snippet in a **notebook code cell**:

```python
import sys
print(sys.executable)
```

If necessary, install using that exact path in the **terminal**. Replace the example path below with the path printed by your notebook.

**Windows — Command Prompt:**

```bat
"C:\path\to\python.exe" -m pip install numpy pandas matplotlib ipykernel
```

**Windows — PowerShell:**

```powershell
& "C:\path\to\python.exe" -m pip install numpy pandas matplotlib ipykernel
```

**macOS / Linux:**

```bash
"/path/to/python" -m pip install numpy pandas matplotlib ipykernel
```

## 4. Restart and test

Return to the notebook, **restart the kernel**, and choose **Run All** to execute all code cells from top to bottom. In VS Code, use the restart button in the notebook toolbar; in JupyterLab, choose **Kernel → Restart Kernel and Run All Cells**. # Part 3. Simple Tests

After installing packages from the terminal in Part 1, restart the notebook kernel and run the following test cells from top to bottom.

## 1. Check Python and imports
This prints your Python version, interpreter path, and package versions. If a package is missing, the import will show an error.

````python
checks = {}  # Reset results before attempting imports.

import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt

print("Python:", sys.version.split()[0])
print("Interpreter:", sys.executable)
print("Working folder:", Path.cwd())
print("NumPy:", np.__version__)
print("pandas:", pd.__version__)
print("Matplotlib:", matplotlib.__version__)
assert 2 + 3 == 5
assert Path.cwd().is_dir()
checks["Python and imports"] = True
print("PASS: Python and imports")
````

## 2. Test NumPy
Create a small array, double its values, and calculate its mean. Expected mean: **3.0**.

````python
numbers = np.array([1, 2, 3, 4, 5])
doubled = numbers * 2
mean = numbers.mean()
print("Doubled:", doubled)
print("Mean:", mean)
assert np.array_equal(doubled, [2, 4, 6, 8, 10])
assert np.isclose(mean, 3.0)
checks["NumPy"] = True
print("PASS: NumPy")
````

## 3. Test pandas
Create a three-row table and calculate the average score. Expected average: **80.0**.

````python
table = pd.DataFrame({
    "Name": ["Alex", "Sam", "Taylor"],
    "Score": [70, 80, 90],
})
average = table["Score"].mean()
print(table)
print("Average score:", average)
assert table.shape == (3, 2)
assert np.isclose(average, 80.0)
checks["pandas"] = True
print("PASS: pandas")
````

## 4. Test Matplotlib
Draw a simple line chart. **Confirm that the chart appears below the cell.** This test does not save any files.

````python
%matplotlib inline
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot([1, 2, 3], [2, 4, 6], marker="o")
ax.set(title="Simple Plot Test", xlabel="x", ylabel="y")
fig.tight_layout()
fig.canvas.draw()
plt.show()
plt.close(fig)
checks["Matplotlib"] = True
print("PASS: plotting code completed. Check that the chart is visible above.")
````

## 5. Final check
After restarting the kernel and running all cells, all four checks should pass with no errors, and **ALL CHECKS PASSED** should appear below. Also confirm that the line chart is visible. Save the notebook after your successful run. Previously saved outputs alone do not verify your current setup. These tests verify basic functionality, not every feature of each package.

````python
expected = ["Python and imports", "NumPy", "pandas", "Matplotlib"]
for name in expected:
    print(f"{'PASS' if checks.get(name) else 'FAIL'}: {name}")
assert all(checks.get(name, False) for name in expected), "Some checks did not pass."
print("\nALL CHECKS PASSED")
````
